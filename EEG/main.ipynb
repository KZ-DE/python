```python
low = 10
high = 30
pole = 4
samp_freq = 250
notch_freq = 50.0
quality_factor = 80

# Desain bandpass filter
def bandpass_filter(data, low, high, fs, order):
    nyq = 0.5 * fs
    low_norm = low / nyq
    high_norm = high / nyq
    b, a = butter(order, [low_norm, high_norm], btype='band')
    return filtfilt(b, a, data)

# Desain notch filter
def apply_notch_filter(data, freq, fs, Q):
    b, a = iirnotch(freq / (0.5 * fs), Q)
    return filtfilt(b, a, data)

# Ambil 1 trial
trial_idx = 0
trial_data = eeg[trial_idx]  # shape (32, 1000)

# Filter semua channel
filtered_data = np.zeros_like(trial_data)

for ch in range(32):
    signal = trial_data[ch]
    signal = apply_notch_filter(signal, notch_freq, samp_freq, quality_factor)
    signal = bandpass_filter(signal, low, high, samp_freq, pole)
    filtered_data[ch] = signal
```

### Domain Frekuensi

```python
low = 0.5
high = 4
pole = 4
samp_freq = 250
notch_freq = 50.0
quality_factor = 80

# Desain bandpass filter
def bandpass_filter(data, low, high, fs, order):
    nyq = 0.5 * fs
    low_norm = low / nyq
    high_norm = high / nyq
    b, a = butter(order, [low_norm, high_norm], btype='band')
    return filtfilt(b, a, data)

# Desain notch filter
def apply_notch_filter(data, freq, fs, Q):
    b, a = iirnotch(freq / (0.5 * fs), Q)
    return filtfilt(b, a, data)

# Ambil 1 trial
trial_idx = 0
trial_data = eeg[trial_idx]  # shape (32, 1000)

# Filter semua channel
filtered_data = np.zeros_like(trial_data)

for ch in range(0,6):
    signal = trial_data[ch]
    signal = bandpass_filter(signal, 0.5, 4, samp_freq, pole)
    filtered_data[ch] = signal
    
for ch in range(6,12):
    signal = trial_data[ch]
    signal = bandpass_filter(signal, 4, 8, samp_freq, pole)
    filtered_data[ch] = signal
    
for ch in range(12,18):
    signal = trial_data[ch]
    signal = bandpass_filter(signal, 8, 13, samp_freq, pole)
    filtered_data[ch] = signal
    
for ch in range(18,24):
    signal = trial_data[ch]
    signal = bandpass_filter(signal, 13, 30, samp_freq, pole)
    filtered_data[ch] = signal
    
for ch in range(24,30):
    signal = trial_data[ch]
    signal = bandpass_filter(signal, 30, 100, samp_freq, pole)
    signal = apply_notch_filter(signal, notch_freq, samp_freq, quality_factor)
    filtered_data[ch] = signal
    
for ch in range(30,32):
    signal = trial_data[ch]
    signal = bandpass_filter(signal, 0.5, 100, samp_freq, pole)
    signal = apply_notch_filter(signal, notch_freq, samp_freq, quality_factor)
    filtered_data[ch] = signal

```
